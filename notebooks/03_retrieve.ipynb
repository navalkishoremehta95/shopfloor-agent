# Retrieve SOP snippets

Query → rank `##` sections by token overlap (lightweight RAG before the LLM).

```python
import sys
from pathlib import Path

root = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
sys.path.insert(0, str(root))

from shopfloor import retrieve

sop = root / "docs/sop_samples/miam_assembly_subset.md"
hits = retrieve(sop, "high risk fastening screwdriver hands clear", top_k=3)
for sid, score, body in hits:
    print(f"--- {sid} ({score:.2f}) ---")
    print(body)
    print()
```
